Part 3: Cleaning Absentee Data

The raw data stored most categorical variables as numbers (day of week was 2 through 6, education was 1 through 4, absence reasons were 0 through 28). Those numbers are meaningless on their own and don't translate well into Tableau. So I created readable label columns by mapping each number to its actual meaning. For example 2 becomes Monday, 1 becomes High School, 23 becomes Medical Consultation. I kept the original numeric columns as well so I didn't lose anything.

I also fixed the Transportation expense column by stripping out commas and converting it to a proper numeric type. After cleaning I re-ran the null check and found 3 nulls in Month Name and 1 in Transportation expense. For Transportation expense I filled it with the median (I chose median over mean because it's more resistant to outliers). For Month Name the nulls came from records where Month of absence was 0, meaning it simply wasn't recorded, so I mapped those to 'Not Recorded' rather than dropping the rows.

```python
df['Transportation expense'] = pd.to_numeric(
    df['Transportation expense'].astype(str).str.replace(',', '').str.strip(),
    errors='coerce'
)

df['Day Name'] = df['Day of the week'].map({
    2: 'Monday', 3: 'Tuesday', 4: 'Wednesday', 5: 'Thursday', 6: 'Friday'
})

df['Season Name'] = df['Seasons'].map({
    1: 'Summer', 2: 'Autumn', 3: 'Winter', 4: 'Spring'
})

df['Education Level'] = df['Education'].map({
    1: 'High School', 2: 'Graduate', 3: 'Postgraduate', 4: 'Master / Doctor'
})

df['Disciplinary Failure'] = df['Disciplinary failure'].map({1: 'Yes', 0: 'No'})

df['Is Drinker'] = df['Drinker'].map({1: 'Drinker',    0: 'Non-Drinker'})
df['Is Smoker']  = df['Smoker'].map( {1: 'Smoker',     0: 'Non-Smoker'})

df['Month Name'] = df['Month of absence'].map({
    1: 'January', 2: 'February', 3: 'March',     4: 'April',
    5: 'May',     6: 'June',     7: 'July',       8: 'August',
    9: 'September', 10: 'October', 11: 'November', 12: 'December'
})

df['Reason Label'] = df['Reason for absence'].map({
    0:  'Unknown',
    1:  'Infectious & Parasitic Diseases',
    2:  'Neoplasms',
    3:  'Blood & Immune Disorders',
    4:  'Endocrine & Metabolic Diseases',
    5:  'Mental & Behavioural Disorders',
    6:  'Diseases of the Nervous System',
    7:  'Diseases of the Eye',
    8:  'Diseases of the Ear',
    9:  'Circulatory Diseases',
    10: 'Respiratory Diseases',
    11: 'Digestive Diseases',
    12: 'Skin Diseases',
    13: 'Musculoskeletal Diseases',
    14: 'Genitourinary Diseases',
    15: 'Pregnancy & Childbirth',
    16: 'Perinatal Conditions',
    17: 'Congenital Malformations',
    18: 'Symptoms & Abnormal Findings',
    19: 'Injury & Poisoning',
    20: 'External Causes',
    21: 'Factors Influencing Health Status',
    22: 'Patient Follow-Up',
    23: 'Medical Consultation',
    24: 'Blood Donation',
    25: 'Laboratory Examination',
    26: 'Unjustified Absence',
    27: 'Physiotherapy',
    28: 'Dental Consultation'
})

label_cols = ['Day Name', 'Season Name', 'Month Name', 'Education Level',
              'Disciplinary Failure', 'Is Drinker', 'Is Smoker', 'Reason Label']
for col in label_cols:
    print(f'   {col}: {df[col].nunique()} unique values')
```

Part 6: Correlation

I used scipy's Pearson correlation to test two numeric variables against absenteeism (number of children and distance from work). I was deliberate about only running correlation on truly numeric continuous variables. The other variables like day of week, education, and drinking status are categorical, so I used averages for those instead, which is simpler and more appropriate.
Children came back with a correlation of +0.114 and a p-value of 0.002, which is statistically significant. Distance came back as -0.088 with a p-value of 0.016, technically significant but I was suspicious of that result because the same employee appears multiple times in the 740 rows. When I looked at the 36 unique employees instead, the p-value jumped to 0.957, meaning the relationship completely disappears. That's an important distinction between statistical significance and practical significance.

```python
from scipy import stats

df_absent = pd.read_csv(r"C:\Users\ewlem\Documents\GitHub\Dekra Assignment\Files\clean_absenteeism.csv")
df_absent.columns = df_absent.columns.str.strip()

comparisons = [
    ('Children',                         'Absenteeism time in hours'),
    ('Distance from Residence to Work',  'Absenteeism time in hours'),
]

print(f'{"Comparison":<55} {"Correlation":>12} {"P-value":>12} {"Significant?":>14}')
print('-' * 97)

for col1, col2 in comparisons:
    clean = df_absent[[col1, col2]].dropna()
    corr, pval = stats.pearsonr(clean[col1], clean[col2])
    sig = 'Yes' if pval < 0.05 else 'No'
    print(f'{col1 + " vs " + col2:<55} {corr:>+12.3f} {pval:>12.4f} {sig:>14}')
```

Part 7: Pulling Consultant Data

The consultant file had a specific structure. Each consultant occupied 3 rows: actual hours, capacity hours, and monthly variance. There was no clean single row per consultant. So I wrote a loop that stepped through every 3 rows, identified the consultant name, and pulled the rolling summary value from the actual row and the capacity row. I then dropped any rows with missing values (those corresponded to consultants Q, R, and T who didn't have complete data) and rounded everything to one decimal place before exporting to Excel for Tableau.

```python
df_consult = pd.read_excel(r"C:\Users\ewlem\Documents\GitHub\Dekra Assignment\Files\Consultant Utilization for Visualization_21.12.xlsx")
output_path  = r"C:\Users\ewlem\Documents\GitHub\Dekra Assignment\Files\consultant_summary.xlsx"
df_consult.columns = df_consult.columns.str.strip()

rows = []
for i in range(1, len(df_consult) - 2, 3):
    consultant = str(df_consult.iloc[i]['Employee']).strip()
    if not (consultant.startswith('Consultant') or consultant.startswith('Mr')):
        continue
    
    actual_row   = df_consult.iloc[i]
    capacity_row = df_consult.iloc[i + 1]
    
    rows.append({
        'Consultant':     consultant,
        'Actual Hours':   actual_row['Rolling Summary'],
        'Capacity Hours': capacity_row['Rolling Summary'],
    })

df_summary = pd.DataFrame(rows)
print(df_summary.to_string())

df_summary.to_excel(output_path, index=False)
print('\nSaved:', output_path)
```

```python
df_summary = df_summary.dropna(subset=['Actual Hours', 'Capacity Hours'])

df_summary['Actual Hours']   = df_summary['Actual Hours'].round(1)
df_summary['Capacity Hours'] = df_summary['Capacity Hours'].round(1)

print(df_summary.to_string())

df_summary.to_excel(output_path, index=False)
print('\nSaved')
```

Part 8: Verifying Employees vs Records

This was important for defending my decision to use all 740 records in the survey analysis rather than averaging down to 36. I grouped the survey data by employee ID and counted the number of unique answers each employee gave across the four survey questions. If an employee always gave the same answer, their max unique count would be 1. If it was greater than 1, their answers changed over time.

The result was 34 out of 36 employees (94%) gave different answers at some point during the year. That validated using all 740 records because each row represents a genuinely different moment in time for that employee, not just a duplicate.

```python
df_clean_survey = pd.read_csv(r"C:\Users\ewlem\Documents\GitHub\Dekra Assignment\Files\clean_survey.csv")

survey_cols = ['HI1_WorkSatisfaction', 'HI2_LikesManager', 'HI3_EnjoysColleagues', 'HI4_WantstoStay']

variation = df_clean_survey.groupby('ID')[survey_cols].nunique()

changed = variation[variation.max(axis=1) > 1]

consistent = variation[variation.max(axis=1) == 1]

print(f'Total unique employees:              {df_clean_survey["ID"].nunique()}')
print(f'Employees whose answers changed:     {len(changed)}')
print(f'Employees who were always consistent:{len(consistent)}')
print()
print('Employees with changing answers:')
print(changed)
```

Part 9: Confirming unique values

I confirmed in Python that 35 of 36 employees have the exact same distance value repeated across every one of their records. Distance is a fixed attribute (it doesn't change between absence events). This proves the 740-record correlation was just repeating the same 36 distance values over and over, which is exactly why it was invalid.

```python
dist_check = df.groupby('ID')['Distance from Residence to Work'].nunique()

all_same = (dist_check == 1).all()
exceptions = dist_check[dist_check > 1]

print(f'Total employees: {df["ID"].nunique()}')
print(f'Employees with same distance repeated across all records: {(dist_check == 1).sum()}')
print(f'Employees with varying distance: {len(exceptions)}')
print(f'All employees have identical distance repeated: {all_same}')

if len(exceptions) > 0:
    print(f'\nExceptions:')
    for emp_id in exceptions.index:
        values = df[df['ID'] == emp_id]['Distance from Residence to Work'].unique()
        print(f'  Employee {emp_id}: {values}')
```
